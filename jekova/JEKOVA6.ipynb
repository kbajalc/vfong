````python
# Integer filters, line by line ports of dsp/lynn.go and dsp/jekova.go (Go). They are NOT used by
# Process any more: it reads the Jekova column the Go pipeline wrote (rec.Jekova), so the two
# can't drift. Kept for plotting and experiments, Jekova gain is 32 as in dsp.NewJekovaFilter.

def Div(a: int, b: int) -> int:
    # Go integer division truncates toward zero, Python // floors
    if b == 0:
        return 0
    q = abs(a) // abs(b)
    return q if (a < 0) == (b < 0) else -q
pass #def

class Lynn:
    # y[n] = 2*y[n-1] - y[n-2] + x[n] - 2*x[n-w/2] + x[n-w], output y[n] / Gain
    def __init__(self, length: int):
        self.Delay = length // 2 - 1
        self.Length = length
        self.Gain = length * length // 4
        self.Reset()
    pass #def

    def Reset(self):
        self.YN = 0
        self.Y1 = 0
        self.Y2 = 0
        self.Value = 0
        self.Index = 0
        self.Buffer = [0] * self.Length
    pass #def

    def Filter(self, datum: int) -> int:
        half = self.Index - self.Length // 2
        if half < 0:
            half += self.Length

        X1 = self.Buffer[half]
        X2 = self.Buffer[self.Index]
        self.Y2 = self.Y1
        self.Y1 = self.YN
        self.YN = 2 * self.Y1 - self.Y2 + datum - 2 * X1 + X2

        self.Buffer[self.Index] = datum
        self.Index += 1
        if self.Index == self.Length:
            self.Index = 0

        self.Value = Div(self.YN, self.Gain)
        return self.Value
    pass #def
pass #class

# Jekova band-pass (paper eq. 1, valid at 250 Hz), multiplied by 8 on both sides:
#     FS[n] = (14*FS[n-1] - 7*FS[n-2] + (x[n] - x[n-2])/2) / 8
#   8*FS[n] = 14*FS[n-1] - 7*FS[n-2] + (x[n] - x[n-2])/2
# YN is 8*FS[n]; the feedback Y1, Y2 is kept at unit gain (YN / JEKOVA_GAIN), the
# output is YN / Gain, so Gain sets the output scale independently of the recursion.
JEKOVA_GAIN = 8

class Jekova:
    def __init__(self, gain: int = 32):
        self.Delay = 0
        self.Gain = gain
        self.Reset()
    pass #def

    def Reset(self):
        self.YN = 0
        self.Y1 = 0
        self.Y2 = 0
        self.X1 = 0
        self.X2 = 0
        self.Value = 0
    pass #def

    def Filter(self, datum: int) -> int:
        self.Y2 = self.Y1
        self.Y1 = Div(self.YN, JEKOVA_GAIN)
        self.YN = 14 * self.Y1 - 7 * self.Y2 + Div(datum - self.X2, 2)

        self.X2 = self.X1
        self.X1 = datum

        self.Value = Div(self.YN, self.Gain)
        return self.Value
    pass #def
pass #class

LYN_WIND = 40 // MS

def LynnFilterInt(x: np.ndarray, w: int = LYN_WIND) -> np.ndarray:
    w -= w % 2
    flt = Lynn(w)
    y = np.array([flt.Filter(int(v)) for v in x])
    return np.roll(y, -flt.Delay)
pass #def

def JakovaFilterInt(x, prep = False) -> tuple[np.ndarray, np.ndarray]:
    y = np.asarray(x, dtype=int)
    y = LynnFilterInt(y)
    ## Jekova Equation ################
    flt = Jekova()
    k = np.array([flt.Filter(int(v)) for v in y])
    ###################################
    return y, k
pass #def
````

````python
# Filters

LYN_WIND = 40 // MS

def LynnFilter(x: np.ndarray, w: int = LYN_WIND) -> np.ndarray:
    w -= w % 2

    a = np.array([1, -2, 1])
    b = np.zeros(w + 1)
    b[0] = 1
    b[w // 2] = -2
    b[w] = 1
    g = (len(b) // 2) ** 2
    shift = w // 2 - 1

    y = signal.lfilter(b, a, x) / g # type: ignore
    y = np.roll(y, -shift)
    return y
pass #def


FS_TARGET = 250    # Hz; every JEKOVA constant is defined for a 250 Hz sampling rate
HP_HZ = 1.0        # first-order high-pass cutoff (per stage)
LP_HZ = 30.0       # Butterworth low-pass cutoff
LP_ORDER = 2       # second-order Butterworth, as the paper specifies
NOTCH_HZ = 50.0    # powerline notch centre frequency (Sofia/European mains)
NOTCH_Q = 30.0     # notch quality factor (dimensionless); bandwidth = NOTCH_HZ / Q,
                   # so Q = 30 gives a ~1.7 Hz notch. The paper does not specify Q.

# Jekova band-pass (paper eq. 1) in integer form, carrying a gain of 16 like the
# Lynn filters. Paper equation, scaled by 16 to clear the fractions:
#     8*FS[i] = 14*FS[i-1] - 7*FS[i-2] + (S[i] - S[i-2])/2
#    16*FS[i] = 28*FS[i-1] - 14*FS[i-2] + (S[i] - S[i-2])
# Coefficients are integers; the peak gain at the 14.6 Hz centre is JEK_GAIN.
# Divide by JEK_GAIN to get unity peak gain. The Step 6 counts are all relative
# to a per-window max / mean / MD, so the cascade is unaffected by the scale.
JEK_B = [16, 0, -16]
JEK_A = [16, -28, 14]

# Gain should be 16, but in integer arithmetics its fails so use 1, the output x16
JEK_GAIN = 16 // 4

def JakovaFilter(x, prep = True) -> tuple[np.ndarray, np.ndarray]:
    y = np.asarray(x, dtype=float)

    if prep:
        ## High pass filter ###############
        hp_b, hp_a = signal.butter(1, HP_HZ / (0.5 * FS_TARGET), btype="highpass") # type: ignore
        y = signal.lfilter(hp_b, hp_a, y)
        y = signal.lfilter(hp_b, hp_a, y)

        ## Low pass filter ################
        lp_b, lp_a = signal.butter(LP_ORDER, LP_HZ / (0.5 * FS_TARGET), btype="lowpass") # type: ignore
        y = signal.lfilter(lp_b, lp_a, y)
        ###################################

        ## Notch filter ###################
        nt_b, nt_a = signal.iirnotch(NOTCH_HZ, NOTCH_Q, fs=FS_TARGET) # type: ignore
        y = signal.lfilter(nt_b, nt_a, y)
    else:
        y = LynnFilter(y)
    pass #if

    ## Jekova Equation ################
    k = signal.lfilter(JEK_B, JEK_A, y) / JEK_GAIN # type: ignore
    ###################################

    return np.array(y), np.array(k)
pass #def
````

````python
# Jekova Algorithm

COLUMNS = [
    "db", 
    "rid", 
    "pon", 
    "pof", 
    "vfb", 

    # "smin",
    # "smax",
    # "savg",
    # "sdev",

    "fmax",
    "favg",
    "mdev",

    "cnt1",
    "cnt2", 
    "cnt3",
    "cnrt",
]

JEKOVA_MIN = 5          # bdc.JEKOVA_MIN, minimal mean |Jekova| of a 1 s block

# The Go pipeline keeps running over the QRX_FLUSH - 1 = 499 padding pushes after the last
# sample, and the Jekova buffer lags Time.Now by 158 samples, so blocks are still closed up to
# 499 - 158 = 341 samples past the end of the record. The padded Jekova output is taken as 0
# (a band-pass of a constant), which approximates what the Go filters produce there.
FLUSH_TAIL = 341

# calculate Jekova counts in a 1 s segment of the band-pass output (paper step 5-6)
def Counts(db, rid, pon, pof, vfb, seg: np.ndarray) -> dict:
    # step 5: the counts are defined on the absolute (rectified) filter output
    abs_fs = np.abs(np.asarray(seg, dtype=float))
    # abs_ly = np.abs(np.asarray(lyn, dtype=float))

    # smin = np.min(lyn)
    # smax = np.max(abs_ly)
    # savg = np.mean(abs_ly)
    # sdev = np.mean(np.abs(abs_ly - savg))   # mean absolute deviation, not median

    fmax = np.max(abs_fs)
    favg = np.mean(abs_fs)
    mdev = np.mean(np.abs(abs_fs - favg))   # mean absolute deviation, not median

    # step 6: number of samples falling in each amplitude band
    cnt1 = int(np.sum(abs_fs >= fmax / 2))                                  # 0.5*smax .. smax
    # Go rule (Signal.VFIB): strictly above the integer-floored mean. The paper's ">= mean" counts
    # every sample of a flat (all zero) block, which would turn a flatline into VF.
    # Signal.VFIB also applies the amplitude floor JEKOVA_MIN: a block with a mean below it is
    # signal loss, not VF, and counts nothing.
    mean = int(abs_fs.sum()) // len(abs_fs)
    cnt2 = int(np.sum(abs_fs > mean)) if mean >= JEKOVA_MIN else 0          # smean    .. smax
    cnt3 = int(np.sum((abs_fs >= favg - mdev) & (abs_fs <= favg + mdev)))   # smean +- MD

    return dict(
        db=db, 
        rid=rid, 
        pon=pon, 
        pof=pof, 
        vfb=vfb, 

        # smin=int(smin), 
        # smax=int(smax), 
        # savg=int(savg),
        # sdev=int(sdev),

        fmax=int(fmax),
        favg=int(favg), 
        mdev=int(mdev),

        cnt1=cnt1, 
        cnt2=cnt2, 
        cnt3=cnt3,

        cnrt = cnt1 * cnt2 / cnt3 if cnt3 > 0 else 0
    )
pass #def

def mean(x: list[float]) -> float:
    return sum(x)
pass #def

AGGREGATE: dict = dict(
    vfb = sum,  # samples of reference VF inside the window

    # smin = min,
    # smax = max,  
    # savg = mean,
    # sdev = mean,

    fmax = max,
    favg = mean,
    mdev = mean,

    cnt1  = sum,  # Jekova step 6 counts
    cnt2  = sum,
    cnt3  = sum,
)

def Window(rows: list[dict], s: int, n: int) -> dict:
    """Aggregate the n consecutive 1 s rows starting at s into one window row."""
    w = rows[s:s + n]
    out = dict(db=w[0]["db"], rid=w[0]["rid"], pon=w[0]["pon"], pof=w[-1]["pof"])
    for name, agg in AGGREGATE.items():
        out[name] = agg(r[name] for r in w)
        if agg == mean:
            out[name] = int(out[name] / n)
        pass #if
    pass #for
    out["cnrt"] = out["cnt1"] * out["cnt2"] / out["cnt3"] if out["cnt3"] > 0 else 0
    return out
pass #def

# epicmp's VF reference (find_episode, type VFE): an episode starts at a '[' (VFON) and ends at
# the next ']' (VFOFF); everything in between, including further '[' and rhythm annotations
# such as (VF / (VFL / (VT, is ignored. An episode with no ']' runs to the end of the record.
def RefBrackets(db: str, rid: str, n: int) -> np.ndarray:
    vfib = np.zeros(n, dtype=int)
    start = None
    with open(f"../work/output/pxg/anr/{db}/{rid}.anr", "r", encoding="utf-8") as f:
        for line in f:
            parts = line.strip().split(",")
            if len(parts) < 2: continue
            if start is None and parts[1] == "[":
                start = int(parts[0])
            elif start is not None and parts[1] == "]":
                vfib[start:int(parts[0])] = 1
                start = None
            pass #if
        pass #for
    pass #with
    if start is not None:
        vfib[start:] = 1
    pass #if
    return vfib
pass #def

def Process(rec: Record, use_butter = False, use_wfon = False) -> tuple[list[dict], list[dict], list[dict], np.ndarray]:
    # use Point for baseline removed, Sensor for original singal
    insig = rec.Point

    # lyn, jek = JakovaFilter(insig, use_butter)
    # rec.Jekova = jek

    jek = np.zeros(((len(insig) + FLUSH_TAIL) // FS) * FS)
    jek[:len(insig)] = rec.Jekova[:len(insig)]

    if use_wfon:
        # epicmp reference: only '[' ... ']' annotations
        vfib = RefBrackets(rec.DB, rec.RID, len(insig))
    else:
        vfib = np.zeros(len(insig), dtype=int)
        for epi in rec.RefEpi:
            if epi.Name in ["#VT", "VFL", "VF", "WF", "VFIB"]:
                vfib[epi.Time:epi.End] = 1
            pass #if
        pass #for
    pass #if
    rec.RefVFB = vfib

    rows = []
    for s in range(len(jek) // FS):
        pon = s * FS
        pof = pon + FS
        vfb = int(np.sum(vfib[pon:pof]))
        row = Counts(rec.DB, rec.RID, pon, pof, vfb, jek[pon:pof])
        rows.append(row)
    pass #for

    row5 = [Window(rows, s, 5) for s in range(len(rows) - 5 + 1)]
    row8 = [Window(rows, s, 8) for s in range(len(rows) - 8 + 1)]

    return rows, row5, row8, rec.RefVFB
pass #def

def Prepare(db: str, use_wfon = False, use_butter = False) -> tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame]:
    rows = []
    row5 = []
    row8 = []

    # mkdir ../work/vfib/{db}
    os.makedirs(f"../work/vfib/{db}", exist_ok=True)

    rids = Rids(db)
    for rid in rids:
        rec = Record(db, rid, wfon=use_wfon)
        rs, r5, r8, vfb = Process(rec, use_butter, use_wfon)
        rows.extend(rs)
        row5.extend(r5)
        row8.extend(r8)

        # save the per-record VFB mask as a compact binary (uint8)
        np.save(f"../work/vfib/{db}/{rid}.ref.npy", vfb.astype(np.uint8))
    pass #for
    dfs = pd.DataFrame(rows, columns=COLUMNS)
    df5 = pd.DataFrame(row5, columns=COLUMNS)
    df8 = pd.DataFrame(row8, columns=COLUMNS)

    # Save to work/vfib/{db}
    dfs.to_csv(f"../work/vfib/{db}_dfs.tsv", index=False, sep="\t", float_format="%.5f")
    df5.to_csv(f"../work/vfib/{db}_df5.tsv", index=False, sep="\t", float_format="%.5f")
    df8.to_csv(f"../work/vfib/{db}_df8.tsv", index=False, sep="\t", float_format="%.5f")

    return dfs, df5, df8
pass #def
````

````python
# Evaluate the published Jekova cascade, thresholds scaled to the window length.

VER = 8

EPOCH = 10 * FS          # the paper's 10 s epoch at 250 Hz = 2500 samples
SPAN = VER * FS          # the 8 s window we use for the cascade, 2000 samples
FRAC = 0.85

# Published thresholds (paper section 2.2.5), valid for a 2500-sample epoch. Every
# count is a number of samples, so all of them scale linearly with the window
# length; the ratio Count1*Count2/Count3 scales linearly too, since N*N/N = N.
JEKOVA_THRESHOLDS = dict(
    C1_LO = 250 * SPAN / EPOCH, 
    C1_HI = 400 * SPAN / EPOCH, 
    C2_LO = 600 * SPAN / EPOCH, 

    C2_HI = 950 * SPAN / EPOCH, 

    C2_MAX = 1100 * SPAN / EPOCH,
    RATIO = 210,
)

def Classify0(wf: pd.DataFrame, thr: dict = JEKOVA_THRESHOLDS) -> np.ndarray:
    # smax = wf["smax"].to_numpy(dtype=float)

    cnt1 = wf["cnt1"].to_numpy(dtype=float)
    cnt2 = wf["cnt2"].to_numpy(dtype=float)
    cnt3 = wf["cnt3"].to_numpy(dtype=float)
    cnrt = wf["cnrt"].to_numpy(dtype=float)

    # r0 = (smax < 0)
    r1 = (cnt1 <  thr["C1_LO"]) & (cnt2 > thr["C2_HI"]) & (cnrt < thr["RATIO"])
    r2 = (cnt1 >= thr["C1_LO"]) & (cnt1 < thr["C1_HI"]) & (cnt2 < thr["C2_LO"]) & (cnrt < thr["RATIO"])
    r3 = (cnt1 >= thr["C1_LO"]) & (cnt2 > thr["C2_HI"])
    r4 = (cnt2 >= thr["C2_MAX"])

    call = np.full(len(wf), "?", dtype="<U1")
    for rule, label in [
            # (r0, "N"),
            (r1, "N"), 
            (r2, "N"), 
            (r3, "S"), 
            (r4, "S"),
        ]:   # first match wins
        call = np.where((call == "?") & rule, label, call)
    pass #for

    return call
pass #def

# The cascade collapses to one threshold on cnt2 (FINDINGS section 1): the c1 gate,
# the ratio and the non-shockable rules never change an output. Same signature and
# labels as Classify, so it drops into EvaluateWin / Evaluate unchanged.
def Classify(wf: pd.DataFrame, thr: dict = JEKOVA_THRESHOLDS) -> np.ndarray:
    return np.where(wf["cnt2"].to_numpy(dtype=float) > thr["C2_HI"], "S", "N")
pass #def
````

````python
# Evaluate 

def Rates(tp: int, fn: int, fp: int, tn: int) -> dict:
    """Confusion counts and derived rates, shared by the window and sample tables."""
    return dict(
        TN=tn, TP=tp, FN=fn, FP=fp, ERR = fp + fn,
        SPC = round(100 * tn / (tn + fp), 2) if tn + fp else float("nan"),
        SEN = round(100 * tp / (tp + fn), 2) if tp + fn else float("nan"),
        PPV = round(100 * tp / (tp + fp), 2) if tp + fp else float("nan"),
        F1  = round(100 * 2 * tp / (2 * tp + fp + fn), 2) if 2 * tp + fp + fn else float("nan"),
    )
pass #def

def MetricsWin(call, shock, nonsh, mixed) -> dict:
    """Window-level confusion for one boolean slice of a database."""
    tp = int(((call == "S") & shock).sum())
    fn = int(((call != "S") & shock).sum())
    fp = int(((call == "S") & nonsh).sum())
    tn = int(((call != "S") & nonsh).sum())
    return dict(nonsh=int((nonsh & ~mixed).sum()), mixed=int(mixed.sum()), shock=int(shock.sum()), **Rates(tp, fn, fp, tn))
pass #def

def EvaluateWin(df: pd.DataFrame) -> pd.DataFrame:
    """Per-record window rows (one per rid) plus a TOTAL row."""
    call = Classify(df)

    # frac is a cut, not a band: every window lands in exactly one class.
    shock = (df["vfb"] >= FRAC * SPAN).to_numpy()
    nonsh = ~shock
    # windows straddling an episode edge: some VF present but under the cut,
    # scored as negatives, reported apart from the clear negatives
    mixed = (df["vfb"].to_numpy() > 0) & nonsh

    db  = df["db"].iloc[0]
    rid = df["rid"].to_numpy()

    wins = []
    for r in pd.unique(rid):
        m = rid == r
        wins.append({"db": db, "rid": str(r), **MetricsWin(call[m], shock[m], nonsh[m], mixed[m])})
    pass #for
    wins.append({"db": db, "rid": "TOTAL", **MetricsWin(call, shock, nonsh, mixed)})
    return pd.DataFrame(wins)
pass #def

# epicmp.c port (VF episode-by-episode comparison, EC57). The reference and test masks are
# turned into episodes (the test mask is still the union of whole 8 s windows called "S"),
# the first 5 minutes are a learning period and are not scored, as in epicmp's default.
# Stats follow epicmp: episode Se / +P (any overlap > 0 is a match) and duration Se / +P.

def Start(db: str) -> int:
    # kit/bxb.go: epicmp gets "-f <skip>" with skip = 0 (DEF_SKIP) for every database but ahads,
    # which is run without -f, i.e. the epicmp default of 5 minutes
    return 5 * 60 * FS if db == "ahads" else 0
pass #def

def Runs(mask: np.ndarray) -> list[tuple[int, int]]:
    """Episodes [start, end) of a boolean mask."""
    m = np.concatenate(([0], np.asarray(mask, dtype=np.int8), [0]))
    d = np.diff(m)
    return list(zip(np.flatnonzero(d == 1).tolist(), np.flatnonzero(d == -1).tolist()))
pass #def

def EpiMatch(ep: list[tuple[int, int]], other: list[tuple[int, int]], start: int, end: int) -> tuple[int, int, int, int]:
    """epicmp(): walk the episodes of one annotator, count those overlapping the other."""
    match = mismatch = dur = ovl = 0
    for s, e in ep:
        if e < start:                 # ends in the learning period
            continue
        if s >= end:
            break
        s = max(s, start)             # begins in the learning period
        e = min(e, end)               # ends after the test period
        d = e - s
        o = sum(max(0, min(e, oe) - max(s, os_)) for os_, oe in other)
        dur += d
        if o > 0:
            match += 1
            ovl += o
        else:
            mismatch += 1
        pass #if
    pass #for
    return match, mismatch, dur, ovl
pass #def

def Pct(a: int, b: int) -> float:
    return int(100.0 * a / b + 0.5) if b > 0 else float("nan")    # epicmp pstat rounding
pass #def

def EpiRow(db: str, rid: str, stp: int, fn: int, ptp: int, fp: int, rdur: int, rov: int, tdur: int, tov: int) -> dict:
    return dict(db=db, rid=rid, STP=stp, FN=fn, PTP=ptp, FP=fp,
                ESe=Pct(stp, stp + fn), EPP=Pct(ptp, ptp + fp),
                DSe=Pct(rov, rdur), DPP=Pct(tov, tdur),
                RefDur=round(rdur / FS, 1), TestDur=round(tdur / FS, 1))
pass #def

def Evaluate(df: pd.DataFrame) -> tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame]:
    """Per-record SAMPLE rows (same columns as EvaluateWin) plus a TOTAL row, and the epicmp table.

    Everything before START (5 min) is a learning period and is not scored.

    The window decisions are mapped back to a per-sample prediction mask (a sample
    is shockable if any covering window was called "S"), compared against the
    per-sample reference, and the mask saved as {db}/{rid}.vfb.npy.
    """

    call = Classify(df)

    # frac is a cut, not a band: every window lands in exactly one class.
    shock = (df["vfb"] >= FRAC * SPAN).to_numpy()
    nonsh = ~shock
    # windows straddling an episode edge: some VF present but under the cut,
    # scored as negatives, reported apart from the clear negatives
    mixed = (df["vfb"].to_numpy() > 0) & nonsh

    db  = df["db"].iloc[0]
    rid = df["rid"].to_numpy()
    pon = df["pon"].to_numpy()
    pof = df["pof"].to_numpy()

    START = Start(db)
    keep = pon >= START

    wins = []
    samp = []
    epis = []
    etot = [0] * 8
    stot = [0, 0, 0, 0]                    # pooled sample TP, FN, FP, TN for TOTAL
    for r in pd.unique(rid):
        m = rid == r
        mk = m & keep
        wins.append({"db": db, "rid": str(r), **MetricsWin(call[mk], shock[mk], nonsh[mk], mixed[mk])})

        ref  = np.load(f"../work/vfib/{db}/{r}.ref.npy").astype(bool)
        pred = np.zeros(len(ref), dtype=bool)
        for a, b, cl in zip(pon[m], pof[m], call[m]):
            if cl == "S":
                pred[a:b] = True
            pass #if
        pass #for
        np.save(f"../work/vfib/{db}/{r}.vfb.npy", pred.astype(np.uint8))

        # Save as detected annotations
        os.makedirs(f"../work/output/pxg/anv/{db}", exist_ok=True)
        with open(f"../work/output/pxg/anv/{db}/{r}.anv", "w") as f:
            prev = -1
            for ix in range(len(pred)):
                p = int(pred[ix])
                if p == prev: continue
                if p == 0:
                    f.write(f"{ix},+,(NF\n")
                else:
                    f.write(f"{ix},+,(WF\n")
                pass #if
                prev = p
            pass #for
            f.write(f"{len(pred)},+,(END\n")
        pass #with

        rs_, ps_ = ref[START:], pred[START:]
        tp = int((ps_ & rs_).sum());  fn = int((~ps_ & rs_).sum())
        fp = int((ps_ & ~rs_).sum()); tn = int((~ps_ & ~rs_).sum())
        for i, v in enumerate((tp, fn, fp, tn)):
            stot[i] += v
        pass #for
        samp.append({"db": db, "rid": str(r), "nonsh": tn + fp, "mixed": 0, "shock": tp + fn, **Rates(tp, fn, fp, tn)})

        # epicmp: reference episodes [start, end], test episodes [start, end)
        rep_, tep_ = Runs(ref), Runs(pred)
        stp, mfn, rdur, rov = EpiMatch(rep_, tep_, START, len(ref))
        ptp, mfp, tdur, tov = EpiMatch(tep_, rep_, START, len(ref))
        epis.append(EpiRow(db, str(r), stp, mfn, ptp, mfp, rdur, rov, tdur, tov))
        for i, v in enumerate((stp, mfn, ptp, mfp, rdur, rov, tdur, tov)):
            etot[i] += v
        pass #for
    pass #for

    wins.sort(key=lambda x: x["ERR"], reverse=True)   # sort by error count, descending

    wins.append({"db": db, "rid": "TOTAL", **MetricsWin(call[keep], shock[keep], nonsh[keep], mixed[keep])})
    tp, fn, fp, tn = stot
    samp.append({"db": db, "rid": "TOTAL", "nonsh": tn + fp, "mixed": 0, "shock": tp + fn, **Rates(tp, fn, fp, tn)})

    epis.append(EpiRow(db, "TOTAL", *etot))

    return pd.DataFrame(wins), pd.DataFrame(samp), pd.DataFrame(epis)
pass #def
````
